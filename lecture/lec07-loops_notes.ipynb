```python
#using loops, can you find the sum of the following
# (0,1,2,3,4,5)
# hint: 0 + 1 + 2 + 3 + 4 + 5 = 15
# print the total sum at the end
# if we are trying to accumulate values, you want to initialize a variable OUTSIDE
# of the for loop
sum = 0 # initialize sum here, OUTSIDE of the for loop to then collect stuff later
#INSIDE the for loop
for i in (0,1,2,3,4,5):
    sum = sum + i #first time, sum = 0, i = 0
    # second time, sum = 0, i = 1; sum = 0 + 1 = 1
    # third time, sum = 1 + 2 = 3,    
    print(sum)
```
