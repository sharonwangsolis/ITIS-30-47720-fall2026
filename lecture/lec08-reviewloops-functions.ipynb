```python
# user-defined function
# let's write our own function that allows us to re-use the code above
# without copying and pasting all those lines of code over and over again

# let's play around with the upper bound number (i < number)
def while_up_to(upper_bound): # function header: def function_name(input_name):
    #this function will print the values of i up until but not including the upper_bound number
    #this is the function body
    i = 0 # initialize i outside of the while loop, different from a for loop
    #where in a for loop, i is a LOCAL variable 
    while i < upper_bound:
        print("loop #", i + 1)
        print("i is", i)
        i = i + 1 # increment i by one at the end of my for loop
        print("i is now", i)
        print("---")
```

```python
def while_up_to(lower_bound, upper_bound): # function header: def function_name(input_name):
    #this function will print the values of i up until but not including the upper_bound number
    #this is the function body
    i = lower_bound # initialize i outside of the while loop, different from a for loop
    #where in a for loop, i is a LOCAL variable 
    loop_number = 0
    while i < upper_bound:
        print("loop #", loop_number+ 1)
        print("i is", i)
        i = i + 1 # increment i by one at the end of my for loop
        loop_number = loop_number + 1
        print("i is now", i)
        print("---")
    return loop_number # alternatively you can return i
```
