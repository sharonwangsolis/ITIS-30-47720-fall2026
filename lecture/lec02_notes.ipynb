```python
# design an algorithm that determines the max in a list of numbers
# step 1: get list of numbers
# step 2: set a placeholder for max seen so far to be the first number in the list
# step 3: for every number in the list, if the number is greater than the max so far,
# set the placeholder for the max to be this new number now. we do this step multiple
# times, to make sure we have looked at every number in the list
# step 4: after we have looked at every single number in our list, we announce that the
# max is the number indicated by the placeholder

# e.g., ( 1, 3, 2)
# step 1: get the list above
# step 2: look at the first number and assign the first number
# as max placeholder, max_placeholder = 1
# step 3: look at next number, and compare with max_placeholder. which is bigger?
# look at 3 and compare with 1, and we see 3 is bigger than 1, 
# and now max_placeholder = 3
# now we look at next number, and compare with max_placeholder, which is bigger?
# look at 2 and compare with max_placeholder, we see 2 is NOT bigger than max_placeholder
# and we leave max_placeholder alone. we are now done with the list
# step 4: announce max  = max_placeholder = 3
```

```python
# psuedocode version 2
# still some english, and some code throughout; think spanglish or code-ish
# let's design an algorithm that determines max in a list of numbers
# step 1: get list of n numbers, n stands for number of numbers x = [1, 3, 2]
# step 2: max = x[0], max is the first item in the list, in python we count starting
# with 0. the first item in a list has an index/position of 0.
# step 3: x[1] > max ? , if so max = x[1], if not do nothing to max
# x[2] > max ?, if so max = x[2], if not do nothing to max
# step 4: print out max
```

```python
# actual python code, remember this is not stuff you should know already
# this is where we're headed
x = [ 1, 3, 2]
# we want to find out max of x
our_max = x[0] # our_max = 1
for i in [0,1,2]: # allows me to check each item in list and iterate steps below
    if x[i] > our_max: # allows me to compare each item in list with max
        our_max = x[i] # update max, since new number is bigger than previous max placeholder
our_max    
```
